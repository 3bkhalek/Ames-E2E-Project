# Remove the outliers & Fill missing values

---



```python
# All numeric columns except ['Year Built', 'Year Remod/Add', 'Garage Yr Blt'], using IQR Method
for column in df.select_dtypes(include=['int64', 'float64']):

  # Skip the 'Year' columns
  if column in ['Year Built', 'Year Remod/Add', 'Garage Yr Blt']:
    continue

  # 4. Impute missing values using the clean dataset's mean
  df[column] = df[column].fillna(df[column].mean())
```

```python
# Check the data after cleaning
df.describe()
```

```python
# Get the categorical columns
categorical_features = [col for col in df.columns if df[col].dtype == 'O']
```

```python
# Impute missing values using the mode method
for col in categorical_features:
  df[col].fillna(df[col].mode()[0], inplace=True)

for col in ['Year Built', 'Year Remod/Add', 'Garage Yr Blt']:
  df[col].fillna(df[col].mode()[0], inplace=True)
```
